```python
import random
import ollama

def iniciar_monitoramento_espacial():

    temp_modulo = random.randint(-30, 130) # Temperatura em Celsius
    nivel_bateria = random.randint(5, 100) # % de energia
    status_sinal = random.choice(["Conexão Estável", "Ruído Magnético Detectado", "Perda Severa de Pacotes"])


    protocolo_seguranca = "Nenhum risco iminente. Módulos operando nominalmente."

    if nivel_bateria <= 20:
        protocolo_seguranca = "CRÍTICO: Reserva de energia baixa. Desligando suporte de vida em setores desocupados."
    elif temp_modulo >= 90:
        protocolo_seguranca = "CRÍTICO: Risco de fusão no núcleo. Ativando resfriamento de emergência."
    elif status_sinal == "Perda Severa de Pacotes":
        protocolo_seguranca = "ATENÇÃO: Dificuldade em contatar a Terra. Tentando reestabelecer via satélite auxiliar."


    dados_telemetria = f"Temperatura: {temp_modulo}°C | Energia: {nivel_bateria}% | Conexão: {status_sinal}. Avalie a situação da missão."


    resposta = ollama.chat(
        model='llama3.2:1b',
        messages=[
            {
                "role": "system",
                "content": "Você é a Inteligência Artificial de bordo de uma Estação Espacial. Sua função é analisar a telemetria e fornecer instruções diretas e curtas para a tripulação se proteger."
            },
            {
                "role": "user",
                "content": dados_telemetria
            }
        ]
    )


    parecer_ia = resposta['message']['content']


    print("=" * 65)
    print(" TERMINAL DE COMANDO - BASE LUNAR ALPHA ")
    print("=" * 65)
    print(f"🌡️  Aquecimento do Núcleo : {temp_modulo}°C")
    print(f"🔋  Capacidade de Bateria : {nivel_bateria}%")
    print(f"📡  Comunicação Terrestre : {status_sinal}")
    print("-" * 65)
    print(f"⚠️  DIAGNÓSTICO DO SISTEMA: {protocolo_seguranca}")
    print("-" * 65)
    print(" DIRETRIZ DA IA DE BORDO:")
    print(parecer_ia)
    print("=" * 65)


iniciar_monitoramento_espacial()
```
